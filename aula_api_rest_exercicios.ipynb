## 2. Fazendo uma requisição GET

O método `GET` é usado para buscar informações em uma API.

Neste exemplo, vamos buscar o post de ID 1.

````python
url_base = "https://jsonplaceholder.typicode.com/"
````

````python
id_post = 6
````

````python
response = requests.get(f"{url_base}/posts/{id_post}")
````

````python
response.status_code
````

````python
response.json()
````

````python
def exibir_post(id_post):
    url_base = "https://jsonplaceholder.typicode.com/"
    response = requests.get(f"{url_base}posts/{id_post}")

    if response.status_code == 200:
        return response.json()
    else:
        return print(f"Falha de requisição. Status code {response.status_code}")
````

````python
exibir_post(50)
````

### Entendendo o código de status

O código `200` indica que a requisição foi bem-sucedida.

Alguns códigos comuns:

| Código | Significado |
|---|---|
| 200 | OK |
| 201 | Created |
| 400 | Bad Request |
| 401 | Unauthorized |
| 404 | Not Found |
| 500 | Internal Server Error |

## 5. Fazendo um POST

O método `POST` é usado para criar um novo recurso.

Neste exemplo, vamos simular a criação de um novo post.

### Interpretação

O código `201` significa **Created**, ou seja, a API aceitou a criação do recurso.

Como essa API é apenas para testes, o novo post não fica realmente salvo na base.

````python
novo_post = {
    "title": "Novo Post",
    "body": "Corpo do nosso novo post. Testando o método HTTP Post",
    "userId": 23
}
````

````python
novo_post
````

````python
response3 = requests.post(f"{url_base}posts", json=novo_post)
response3.status_code
````

````python
def criar_post(title, body):
    response = requests.post(f"https://jsonplaceholder.typicode.com/posts/", json= {
        "title": title,
        "body": body,
        "userId": 1
    })
    return response.json()

criar_post("teste", "teste")
````

## 6. Fazendo um PUT

O método `PUT` é usado para atualizar um recurso inteiro.

Neste exemplo, vamos simular a atualização do post de ID 1.

````python
{
    'userId': 5,
    'id': 1,
    'title': 'titulo aleatorio',
    'body': 'novo body'
}
````

````python
url_base
````

````python
put_response = requests.put(f'{url_base}posts/50', json = {
    'userId': 5,
    'id': 1,
    'title': 'titulo aleatorio',
    'body': 'novo body'
})

put_response.status_code
````

````python
exibir_post(50)
````

````python

def update_post_by_id(id, body):
    url_base = "https://jsonplaceholder.typicode.com/"
    response = requests.get(f"{url_base}/posts/{id}")

    if response.status_code == 200:
        dados_post = response.json()

    dados_post["id"] = id
    dados_post["body"] = body

    put_response = requests.put(f"{url_base}posts/{id}", json = dados_post)

    if put_response.status_code == 200:
        return put_response.json()
    else:
        print(f"Falha ao alterar o body. Status code: {put_response.status_code}")

resposta = update_post_by_id(50, "funcionou!")
````

````python
resposta
````

## 7. Fazendo um PATCH

O método `PATCH` é usado para atualizar apenas parte de um recurso.

A diferença principal é:

- `PUT`: geralmente envia o objeto inteiro atualizado;
- `PATCH`: envia apenas os campos que serão alterados.

````python
def patch_title(title, id):
    url_base = "https://jsonplaceholder.typicode.com/"
    response = requests.patch(f"{url_base}posts/{id}", json = {
        'title': title
    })

    if response.status_code == 200:
        return response.json()
    else:
        return print(f"Falha ao alterar o body. Status code: {put_response.status_code}")
````

## 8. Fazendo um DELETE

O método `DELETE` é usado para excluir um recurso.

Neste exemplo, vamos simular a exclusão do post de ID 1.

````python
def delete_by_id(id):
    url_base = "https://jsonplaceholder.typicode.com/"
    response = requests.delete(f"{url_base}/posts/{id}")

    return response.status_code

delete_by_id(5)
````
